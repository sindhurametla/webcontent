```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from scraper import fetch_website_contents
```

```python
load_dotenv()
```

```python
prompt = ChatPromptTemplate.from_template(            # ①
    "Give a short 2 line, friendly summary of this website:\n\n{website}")

model = ChatGroq(model = "openai/gpt-oss-120b", temperature = 0.3)

parser = StrOutputParser()

chain = prompt | model | parser

def summerize(url):
    return chain.invoke({"website": fetch_website_contents(url)})

print(summerize("https://anthropic.com"))
```

```python
# LLMs are stateless, so "memory" means sending past messages with every request.
# MessagesPlaceholder is the slot where that list of past messages gets inserted.
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a friendly tutor."),     # ① sets the model's role and tone
    MessagesPlaceholder("history"),              # ② past turns park here
    ("human", "{question}"),                     # ③ the new question
])
chain = prompt | model

# Hardcoded history for demonstration. In a real chat loop, append each new
# HumanMessage and AIMessage to this list after every turn.
history = [HumanMessage("My name is Sindhura."), AIMessage("Hi Sindhura!")]

# The model can answer because the earlier turns are included in the prompt,
# not because it remembers anything on its own.
print(chain.invoke({"history": history, "question": "What's my name?"}).content)
```
